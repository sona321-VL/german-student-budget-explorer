```python

# German Student Budget Explorer 🇩🇪
# Step 1: Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("German Student Budget Explorer started!")
print("All libraries imported successfully.")
```

```python

# Step 2: Create the student budget dataset

data = {
    "City": ["Berlin", "Munich", "Hamburg", "Aachen"],
    "Rent": [450, 650, 500, 400],
    "Food": [250, 280, 250, 230],
    "Transport": [50, 60, 55, 50],
    "Health_Insurance": [140, 140, 140, 140],
    "Phone": [15, 15, 15, 15],
    "Personal_Expenses": [100, 120, 100, 90]
}

df = pd.DataFrame(data)

# Display the dataset
df
```

```python

# Step 3: Calculate total monthly expenses

expense_columns = [
    "Rent",
    "Food",
    "Transport",
    "Health_Insurance",
    "Phone",
    "Personal_Expenses"
]

df["Total_Monthly_Expenses"] = df[expense_columns].sum(axis=1)

df[["City", "Total_Monthly_Expenses"]]
```

```python

# Step 4: Create our first bar chart

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    df["City"],
    df["Total_Monthly_Expenses"]
)

plt.title("Monthly Student Expenses in German Cities")
plt.xlabel("City")
plt.ylabel("Total Monthly Expenses (€)")

plt.tight_layout()
plt.show()
```

```python

# Step 5: Save the chart

plt.figure(figsize=(10, 6))

plt.bar(
    df["City"],
    df["Total_Monthly_Expenses"]
)

plt.title("Monthly Student Expenses in German Cities")
plt.xlabel("City")
plt.ylabel("Total Monthly Expenses (€)")

plt.tight_layout()

plt.savefig(
    "../visualizations/monthly_budget.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Chart saved successfully!")
```

```python

# Step 6: Analyze average expenses by category

expense_columns = [
    "Rent",
    "Food",
    "Transport",
    "Health_Insurance",
    "Phone",
    "Personal_Expenses"
]

average_expenses = df[expense_columns].mean().sort_values(
    ascending=False
)

print("Average monthly expenses by category:")
display(average_expenses)
```

```python

# Step 7: Visualize average expenses by category

plt.figure(figsize=(10, 6))

average_expenses.plot(
    kind="bar",
    title="Average Monthly Student Expenses by Category",
    xlabel="Expense Category",
    ylabel="Average Expense (€)"
)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig(
    "../visualizations/average_expenses.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python

# Step 8: Calculate average expense percentages

total_average_expense = average_expenses.sum()

expense_percentage = (
    average_expenses / total_average_expense
) * 100

expense_summary = pd.DataFrame({
    "Average Expense (€)": average_expenses,
    "Percentage (%)": expense_percentage
})

display(expense_summary.round(2))
```

```python

# Step 9: Visualize the budget distribution

plt.figure(figsize=(8, 8))

plt.pie(
    average_expenses,
    labels=average_expenses.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Average Student Expenses")
plt.tight_layout()

plt.savefig(
    "../visualizations/budget_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# Step 10: Create a student budget calculator

monthly_budget = 1200  # Example budget in euros

df["Remaining_Budget"] = (
    monthly_budget - df["Total_Monthly_Expenses"]
)

df["Budget_Status"] = df["Remaining_Budget"].apply(
    lambda amount: "Within Budget" if amount >= 0
    else "Over Budget"
)

display(
    df[
        [
            "City",
            "Total_Monthly_Expenses",
            "Remaining_Budget",
            "Budget_Status"
        ]
    ]
)
```

```python

# Step 11: Compare remaining budgets

plt.figure(figsize=(9, 5))

plt.bar(
    df["City"],
    df["Remaining_Budget"]
)

plt.axhline(y=0, linestyle="--")

plt.title("Monthly Budget Remaining by German City")
plt.xlabel("City")
plt.ylabel("Remaining Budget (€)")

plt.tight_layout()

plt.savefig(
    "../visualizations/remaining_budget.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
